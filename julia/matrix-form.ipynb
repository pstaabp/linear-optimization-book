```julia
include("SimplexTableau.jl")
using LinearAlgebra, Latexify, .SimplexTableau
```

```julia
A = [-18 -5 -10
  -4 -20 -15
  10 15 18
  20 16 4]
  b = [-85, -70, 250, 180]
  c = [40, 36, 72]
```

```julia
β = [1,3,6,7]
π = setdiff(collect(1:7), β)
```

```julia
B = rationalize.(hcat(A, I)[:, β])
Π = hcat(A, I)[:, π]
```

```julia
c = [40, 36, 72, 0, 0, 0, 0]
```

```julia
inv(B)
```

```julia
inv(B)*b
```

```julia
inv(B)*Π
```

```julia
c[β]'*inv(B)*b
```

```julia
c[β]'*inv(B)*Π -c[π]'
```

```julia
dβ = round(Int, det(B))
```

```julia
Bp = round.(Int, dβ * inv(B))
```

```julia
Bp*b
```

```julia
Bp*Π
```

```julia
c[β]'*Bp*b
```

```julia
c[β]'*Bp*Π - dβ *c[π]'
```

## Simplex method on the above problem

```julia
β = [1,3,4,5]
π = setdiff(collect(1:7), β)
```

```julia
(B = rationalize.(hcat(A,I)[:,β]) ) |> latexify
```

```julia
(Π = hcat(A,I)[:,π] ) |> latexify
```

```julia
dβ = abs(round(Int, det(B)))
```

```julia
(Bp = round.(Int, dβ * inv(B)) ) |> latexify
```

```julia
c[β]'*Bp*Π-dβ*c[π]'
```

```julia
Bp*b |> latexify
```

```julia
Bp*Π |> latexify
```

```julia
Bp*Π[:,2]
```

```julia
 (Bp*b) ./ (Bp*Π[:,2])
```
