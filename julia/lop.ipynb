```julia
using JuMP, HiGHS
```

```julia
m = Model(HiGHS.Optimizer)
@variable(m, x₁ ≥ 0)
@variable(m, x₂ ≥ 0)
@constraint(m, c1, 4x₁ + 3x₂ ≤ 120)
@constraint(m, c2, x₁+2x₂ ≤40)
@constraint(m, c3, x₂ ≤ 16)
@objective(m, Max, 2x₁+3x₂)
print(m)
```

```julia
optimize!(m)
```

```julia
is_solved_and_feasible(m)
```

```julia
value(x₁),value(x₂), objective_value(m)
```

```julia
get_attribute.([x₁,x₂], MOI.VariableBasisStatus())
```

```julia
get_attribute.([c1,c2,c3], MOI.ConstraintBasisStatus())
```

```julia
dual(c1), dual(c2), dual(c3)
```

```julia
m2 = Model(HiGHS.Optimizer)
@variable(m2, y₁ ≥ 0)
@variable(m2, y₂ ≥ 0)
@variable(m2, y₃ ≥ 0)
@constraint(m2, 5y₁ + 18y₂ + 10y₃ ≥ 54)
@constraint(m2, 7y₁ + 10y₂ + 24y₃ ≥ 60)
@constraint(m2, 4y₁ + 8y₂ + 12y₃ ≥ 45)
@objective(m2, Min, 600y₁ + 750y₂ + 500y₃ )
print(m2)
```

```julia
set_silent(m2)
optimize!(m2)
is_solved_and_feasible(m2)
```

```julia
value(y₁),value(y₂), value(y₃), objective_value(m2)
```

```julia
 solution_summary(m2)
```

```julia
 solution_summary(m2, verbose = true)
```

## An infeasible problem

```julia
m4 = Model(HiGHS.Optimizer)
@variable(m4, x₁ ≥ 0)
@variable(m4, x₂ ≥ 0)
@constraint(m4, x₁ - 2x₂ ≤ -6)
@constraint(m4, -5x₁ + 2x₂ ≤ -20)
@constraint(m4, -3x₁ -5x₂ ≤ -75)
@objective(m4, Max, 5x₁ + 3x₂)
print(m4)
```

```julia
set_silent(m4)
optimize!(m4)
is_solved_and_feasible(m4)
```

```julia
solution_summary(m4)
```

## Unbounded Problem

```julia
m5 = Model(HiGHS.Optimizer)
@variable(m5, x₁ ≥ 0)
@variable(m5, x₂ ≥ 0)
@constraint(m5, -x₁ + 2x₂ ≤ 6)
@constraint(m5, 5x₁ - 2x₂ ≤ 20)
@constraint(m5, -3x₁ - 5x₂ ≤ -75)
@objective(m5, Max, 5x₁ + 3x₂)
print(m5)
```

```julia
set_silent(m5)
optimize!(m5)
is_solved_and_feasible(m5)
```

```julia
solution_summary(m5)
```
