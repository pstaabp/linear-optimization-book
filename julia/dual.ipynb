```julia
using JuMP, HiGHS
```

```julia
m = Model(HiGHS.Optimizer)
@variable(m, x₁ ≥ 0)
@variable(m, x₂ ≥ 0)
@variable(m, x₃ ≥ 0)
@objective(m, Max, 3x₁ + 4x₂ + 7x₃)
@constraint(m, c1, x₁ + 2x₂ + 3x₃ ≤ 12)
@constraint(m, c2, 4x₁ + 3x₂ + 2x₃ ≤ 18)
print(m)
```

```julia
set_silent(m)
optimize!(m)
is_solved_and_feasible(m)
```

$$\begin{aligned}
\text{Maximize} && z = 15x_1+20x_2-5x_3-7x_4,& \\
\text{subject to} && x_1 +2x_2-3x_3 & \leq 10, \\
&& 3x_1 +x_2+4x_4 \leq 15,& \\
&& 2x_1 - 3x_2 -x_3+4x_4 \leq 5, \\
\text{and}&&x_1,x_2,x_3,x_4 \geq 0.
\end{aligned}
$$

1. Solve the LOP and determine if it has a solution.
2. If so, find the values of 4 original equations, the constraint values and the objective value. 
3. Find the values of the slack variables $x_5, x_6$ and $x_7$.
3. Find all basic and nonbasic variables.
4. Find the certificate (the entire objective row of the final tableau.)
5. Use the certificate to find the objective value of the dual problem.  That is, find $w = \boldsymbol{b}^{\intercal} \boldsymbol{y}$

```julia
m2 = Model(HiGHS.Optimizer)
@variable(m2, x₁ ≥ 0)
@variable(m2, x₂ ≥ 0)
@variable(m2, x₃ ≥ 0)
@variable(m2, x₄ ≥ 0)
@objective(m2, Max, 15x₁ + 20x₂ - 5x₃ - 7x₄)
@constraint(m2, c1, x₁ + 2x₂ - 3x₃ ≤ 10)
@constraint(m2, c2, 3x₁ + x₂ + 4x₄ ≤ 15)
@constraint(m2, c3, 2x₁ - 3x₂ - x₃ + 4x₄  ≤ 5)
print(m2)
```

```julia
set_silent(m2)
optimize!(m2)
is_solved_and_feasible(m2)
```
