```julia
using JuMP, HiGHS
```

```julia
m = Model(HiGHS.Optimizer)
@variable(m, x₁ ≥ 0)
@variable(m, x₂ ≥ 0)
@variable(m, x₃ ≥ 0)
@objective(m, Max, 3x₁ + 4x₂ + 7x₃)
@constraint(m, c1, x₁ + 2x₂ + 3x₃ ≤ 12)
@constraint(m, c2, 4x₁ + 3x₂ + 2x₃ ≤ 18)
print(m)
```

```julia
set_silent(m)
optimize!(m)
is_solved_and_feasible(m)
```

```julia
value(x₁), value(x₂), value(x₃)
```

```julia
objective_value(m)
```

```julia
md = Model(HiGHS.Optimizer)
@variable(md, y₁ ≥ 0)
@variable(md, y₂ ≥ 0)
@objective(md, Min, 12y₁ + 18y₂)
@constraint(md, cd1, y₁ + 4y₂ ≥ 3)
@constraint(md, cd2, 2y₁ + 3y₂ ≥ 4)
@constraint(md, cd3, 3y₁ + 2y₂ ≥ 7)
print(md)
```

```julia
set_silent(md)
optimize!(md)
is_solved_and_feasible(md)
```

```julia
value(y₁), value(y₂)
```

```julia
objective_value(md)
```
